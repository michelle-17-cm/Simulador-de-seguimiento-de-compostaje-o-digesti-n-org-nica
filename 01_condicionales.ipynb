# Integrante 1: Condicionales en el seguimiento de compostaje

Los condicionales (`if`, `elif`, `else`) permiten tomar decisiones según si una condición se cumple o no. En el seguimiento de compostaje sirven para evaluar variables como temperatura, humedad, pH, oxígeno o relación C/N y mostrar alertas cuando el proceso requiere intervención[cite: 4].va.

```python
temperatura = 45
humedad = 55.0
relacion_cn = 28.0
oxigeno = 16.5
ph = 6.8

if temperatura < 35:
    print("ALERTA: Temperatura baja (actividad lenta).")
elif temperatura <= 65:
    print("OK: Temperatura en rango operativo (35 °C - 65 °C).")
else:
    print("ALERTA CRITICA: Sobrecalentamiento (> 65 °C).")

if humedad < 45:
    print("ALERTA: Humedad baja (< 45%). Se recomienda regar.")
elif humedad > 65:
    print("ALERTA: Humedad alta (> 65%). Incorporar estructurante seco.")
else:
    print("OK: Humedad adecuada (45% - 65%).")

if oxigeno < 10:
    print("ALERTA: Oxigeno bajo (< 10%). Realizar volteo o aireacion.")
else:
    print("OK: Oxigeno adecuado (>= 10%).")

if relacion_cn < 20 or relacion_cn > 35:
    print("ALERTA: Relacion C/N fuera de rango (20 - 35).")
else:
    print("OK: Relacion C/N adecuada.")

if ph < 6.0 or ph > 8.5:
    print("ALERTA: pH fuera de rango (6.0 - 8.5).")
else:
    print("OK: pH adecuado.")
```
