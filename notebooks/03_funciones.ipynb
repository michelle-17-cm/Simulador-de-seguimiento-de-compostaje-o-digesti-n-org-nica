```python
# SIMULADOR DE SEGUIMIENTO DE COMPOSTAJE
# Integrante responsable: Funciones

print("SIMULADOR DE COMPOSTAJE")
print("Módulo de funciones")
print("------------------------------")
```

```python
def calcular_promedio(datos):
    if len(datos) == 0:
        return 0

    promedio = sum(datos) / len(datos)
    return promedio
```

```python
temperaturas = [35, 40, 45, 50, 55]

resultado = calcular_promedio(temperaturas)

print("Temperaturas registradas:", temperaturas)
print("Temperatura promedio:", resultado, "°C")
```

```python
def calcular_humedad(datos):
    if len(datos) == 0:
        return 0

    humedad_promedio = sum(datos) / len(datos)
    return humedad_promedio
```

```python
def calcular_variacion(datos):
    if len(datos) < 2:
        return 0

    variacion = datos[-1] - datos[0]
    return variacion
```

```python
def resumen_compostaje(temperaturas, humedades):

    promedio_temp = calcular_promedio(temperaturas)
    promedio_humedad = calcular_humedad(humedades)
    variacion_temp = calcular_variacion(temperaturas)

    print("===== RESUMEN DEL COMPOSTAJE =====")
    print("Temperatura promedio:", promedio_temp, "°C")
    print("Humedad promedio:", promedio_humedad, "%")
    print("Variación de temperatura:", variacion_temp, "°C")
```

```python
temperaturas = [35, 40, 45, 50, 55]
humedades = [50, 55, 60, 58, 62]

resumen_compostaje(temperaturas, humedades)
```
